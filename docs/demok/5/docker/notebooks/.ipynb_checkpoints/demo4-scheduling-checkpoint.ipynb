## 1. Cron szintaxis és Airflow shortcut-ok

Az Airflow `schedule` paramétere elfogad:
- **Cron string**: `'0 6 * * *'` = minden nap 06:00
- **Shortcut**: `'@daily'` = `'0 0 * * *'`
- **timedelta**: `timedelta(hours=4)` = 4 óránként
- **`None`**: csak manuálisan triggelhető (sensor-vezérelt pipeline-oknál)

```
┌───────────── perc  (0–59)
│ ┌─────────── óra   (0–23)
│ │ ┌───────── nap   (1–31)
│ │ │ ┌─────── hónap (1–12)
│ │ │ │ ┌───── hét napja (0–7, 0 és 7 = vasárnap)
│ │ │ │ │
* * * * *
```

````python
# Cron kifejezések és értelmezésük
schedules = [
    ('0 6 * * *',    'Minden nap 06:00'),
    ('0 8 * * 1-5',  'Hétköznap 08:00'),
    ('0 0 1 * *',    'Minden hónap 1-je 00:00'),
    ('0 */4 * * *',  '4 óránként'),
    ('*/15 * * * *', '15 percenként'),
    ('@daily',       '= 0 0 * * * (minden nap éjfél)'),
    ('@weekly',      '= 0 0 * * 0 (vasárnap éjfél)'),
    ('@monthly',     '= 0 0 1 * * (hónap eleje)'),
]

print(f'{'Cron':<20} {'Jelentés'}')
print('-' * 50)
for cron, desc in schedules:
    print(f'{cron:<20} {desc}')
````

## 2. Idempotencia – a pipeline újrafuttathatóság alapja

**Idempotens task**: bármennyi újrafuttatás esetén ugyanazt az eredményt adja.
Ez a retry safety alapja: ha egy task megbukik és újraindul, nem okoz duplikált adatot.

**Stratégiák:**
1. `DELETE + INSERT`: töröljük az adott napot, majd újraszúrjuk
2. `UPSERT (INSERT OR REPLACE)`: ha létezik, frissíti; ha nem, szúrja be
3. **Atomikus swap**: staging → production táblacsere

````python
# Demonstráció: nem idempotens vs idempotens
con = duckdb.connect()

# Setup: orders tábla
con.execute("CREATE TABLE orders_bad  (date VARCHAR, revenue FLOAT)")
con.execute("CREATE TABLE orders_good (date VARCHAR PRIMARY KEY, revenue FLOAT)")

ds = '2024-01-15'

# NEM IDEMPOTENS: minden futásnál új sort szúr be
def bad_insert(ds):
    con.execute(f"INSERT INTO orders_bad VALUES ('{ds}', 1234.56)")

# IDEMPOTENS: DELETE + INSERT
def good_delete_insert(ds):
    con.execute(f"DELETE FROM orders_good WHERE date = '{ds}'")
    con.execute(f"INSERT INTO orders_good VALUES ('{ds}', 1234.56)")

# Mindkét task 3x fut (szimulált retry)
for _ in range(3):
    bad_insert(ds)
    good_delete_insert(ds)

bad_count  = con.execute(f"SELECT COUNT(*) FROM orders_bad  WHERE date='{ds}'").fetchone()[0]
good_count = con.execute(f"SELECT COUNT(*) FROM orders_good WHERE date='{ds}'").fetchone()[0]

print(f'Nem idempotens (3x futás): {bad_count} sor – DUPLIKÁTUM!')
print(f'Idempotens     (3x futás): {good_count} sor – helyes')
````

## 3. Backfill – historikus adatok visszatöltése

Ha egy DAG-ot több napig nem futtatták, a `backfill` paranccsal az összes kimaradt
napra manuálisan futtathatunk:

```bash
# Január összes napjára visszamenőleg
airflow dags backfill \
    --dag-id daily_revenue_load \
    --start-date 2024-01-01 \
    --end-date   2024-01-31
```

**catchup=False** (ajánlott): a kimaradt futások **nem** pótlódnak automatikusan.
**catchup=True**: a kimaradt futások mind egyszerre indulnak → Run lavina veszélye!

**Best practice:** mindig `catchup=False` + explicit `backfill` parancs, kontrollált visszatöltéshez.

````python
# Backfill szimulálása Python-ban (Airflow nélkül)
con = duckdb.connect()
con.execute("""
    CREATE TABLE daily_revenue (
        date    VARCHAR PRIMARY KEY,
        revenue FLOAT
    )
""")

def load_day(ds: str, revenue: float) -> None:
    """Idempotens napi betöltés."""
    con.execute('DELETE FROM daily_revenue WHERE date = ?', [ds])
    con.execute('INSERT INTO daily_revenue VALUES (?, ?)', [ds, revenue])

# Backfill: január 1–10.
print('Backfill futtatása (2024-01-01 → 2024-01-10):')
start = date(2024, 1, 1)
for i in range(10):
    ds  = (start + timedelta(days=i)).isoformat()
    rev = round(random.uniform(100000, 500000), 2)
    load_day(ds, rev)
    print(f'  {ds}: {rev:,.0f}')

total = con.execute('SELECT COUNT(*), SUM(revenue) FROM daily_revenue').fetchone()
print(f'\nBetöltött napok: {total[0]}, összesen: {total[1]:,.0f}')
````

## 4. execution_date vs futási idő

**Fontos Airflow fogalom:** az `execution_date` (más néven `ds`) a Run **logikai időpontja**,
nem az a pillanat, amikor ténylegesen fut.

Példa: `@daily` DAG, `start_date=2024-01-01`
- A `2024-01-02` napra szóló Run `execution_date`-je: **2024-01-01**
- Ez azt jelzi, hogy a 2024-01-01 napi adatokat kell feldolgozni
- A Run tényleges futási ideje: 2024-01-02 00:00

**Ezért veszélyes `start_date=datetime.now()`:** a Scheduler újra-parseoláskor mindig más értéket lát!

````python
# execution_date logika demonstrálása
from datetime import datetime, timedelta

start_date = datetime(2024, 1, 1)
schedule_interval = timedelta(days=1)  # @daily

print('Airflow run ütemezés – @daily, start_date=2024-01-01:')
print(f'{'execution_date (ds)':<25} {'Tényleges futási idő':<25} {'Feldolgozandó adat'}')
print('-' * 75)

for i in range(5):
    execution_date = start_date + i * schedule_interval
    actual_run     = start_date + (i + 1) * schedule_interval  # 1 periódussal később fut
    data_period    = execution_date.strftime('%Y-%m-%d')         # ezt a napot dolgozza fel
    print(f'{execution_date.strftime("%Y-%m-%d"):<25} '
          f'{actual_run.strftime("%Y-%m-%d"):<25} '
          f'{data_period}')
````
