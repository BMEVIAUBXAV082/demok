## 1. @flow és @task – az alapelemek

- **`@flow`**: orchestration egység – log, monitoring, retry logika
- **`@task`**: újrafelhasználható munkaegység – cache-elhető, retry-olható
- A flow-k egymásba ágyazhatók (subflow)
- Nincs szükség DAG fájlra – a Python maga a definíció

````python
from prefect import flow, task
from prefect.tasks import task_input_hash
from datetime import timedelta
import random, hashlib

@task(retries=2, retry_delay_seconds=5,
      log_prints=True)
def extract_data(date: str) -> list[dict]:
    """Szimulált API hívás – napi adatok letöltése."""
    random.seed(int(hashlib.md5(date.encode()).hexdigest(), 16) % 1000)
    records = [{'id': i, 'amount': round(random.uniform(100, 50000), 2),
               'date': date}
               for i in range(1, 31)]
    print(f'Letöltve: {len(records)} rekord ({date})')
    return records

@task(cache_key_fn=task_input_hash,
      cache_expiration=timedelta(hours=1),
      log_prints=True)
def validate_data(records: list[dict]) -> list[dict]:
    """Validáció – azonos inputra cache-ből tér vissza."""
    valid = [r for r in records if r['amount'] > 0]
    print(f'Validált: {len(valid)} / {len(records)}')
    return valid

@task(log_prints=True)
def load_data(records: list[dict]) -> int:
    total = sum(r['amount'] for r in records)
    print(f'Betöltve: {len(records)} sor | Összeg: {total:,.0f}')
    return len(records)

@flow(name='etl-pipeline', log_prints=True)
def etl_pipeline(date: str = '2024-01-15') -> int:
    raw   = extract_data(date)
    valid = validate_data(raw)
    return load_data(valid)

# Közvetlen futtatás – Scheduler nélkül
result = etl_pipeline(date='2024-01-15')
print(f'\nFlow eredmény: {result} sor')
````

## 2. Cache – azonos inputra nem fut újra

`cache_key_fn=task_input_hash`: az input adatokból hash-t számol.
Ha ugyanazzal az inputtal hívják meg a task-ot és a cache még érvényes → az előző eredményt adja vissza.

**Mikor hasznos?**
- Drága API hívások (pl. ML feature extraction)
- Idempotens transzformációk: ha a forrás nem változott, az eredmény sem változott
- Fejlesztés közben: gyorsabb iteráció

````python
# Cache demonstráció – 2x ugyanaz az input
import time

print('1. hívás:')
t0 = time.time()
r1 = etl_pipeline(date='2024-02-01')
t1 = time.time()
print(f'  Idő: {t1-t0:.2f}s')

print('\n2. hívás (azonos dátum – cache hit):')
t2 = time.time()
r2 = etl_pipeline(date='2024-02-01')
t3 = time.time()
print(f'  Idő: {t3-t2:.2f}s')

print('\n3. hívás (más dátum – cache miss):')
t4 = time.time()
r3 = etl_pipeline(date='2024-02-02')
t5 = time.time()
print(f'  Idő: {t5-t4:.2f}s')
````

## 3. Subflow – moduláris pipeline

A Prefect flow-k egymásba ágyazhatók: egy `@flow` meghívhat más `@flow`-kat.
Ez lehetővé teszi a moduláris pipeline-tervezést: minden üzleti folyamat saját flow-ban,
a master flow koordinálja őket.

````python
from prefect import flow, task

@flow(name='extract-subflow')
def extract_subflow(date: str) -> list[dict]:
    return [{'id': i, 'amount': 1000.0, 'date': date} for i in range(1, 11)]

@flow(name='transform-subflow')
def transform_subflow(records: list[dict]) -> list[dict]:
    return [{'id': r['id'], 'amount_huf': r['amount'] * 390} for r in records]

@flow(name='master-etl-pipeline')
def master_pipeline(dates: list[str]) -> dict:
    """Master flow: több napot dolgoz fel szubflow-kkal."""
    results = {}
    for date in dates:
        raw         = extract_subflow(date)        # subflow hívás
        transformed = transform_subflow(raw)       # subflow hívás
        results[date] = len(transformed)
    return results

summary = master_pipeline(['2024-01-01', '2024-01-02', '2024-01-03'])
print('Összesítő:', summary)
````

## 4. Deployment és ütemezés

```bash
# Prefect szerver indítása
prefect server start
# UI: http://localhost:4200

# Deployment létrehozása
prefect deploy --name etl-daily --cron '0 3 * * *'

# Work Pool worker indítása (ez futtatja a flow-kat)
prefect worker start --pool default-agent-pool
```

**Work Pool** = worker-ek gyűjteménye, amelyek a deployment-eket futtatják.
Típusai: `process` (helyi), `docker`, `kubernetes` – a deployment-ben megadható.

````python
# Deployment konfiguráció YAML-ban (prefect.yaml)
deployment_yaml = '''
name: etl-daily
entrypoint: demo2_prefect_flows.py:etl_pipeline
schedules:
  - cron: "0 3 * * *"
    timezone: "Europe/Budapest"
    active: true
work_pool:
  name: default-agent-pool
  work_queue_name: default
parameters:
  date: "{{ run_config.date }}"
'''
print('Deployment YAML:')
print(deployment_yaml)
````
