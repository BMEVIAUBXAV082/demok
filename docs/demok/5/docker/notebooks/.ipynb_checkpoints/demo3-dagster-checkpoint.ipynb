## 1. @asset – az alap building block

**Software-Defined Asset (SDA):** az asset definiálja, hogy:
- **mit** állít elő (egy DataFrame, tábla, fájl)
- **hogyan** állítja elő (a függvény törzse)
- **mikor kell frissíteni** (Freshness Policy)
- **milyen más asset-ektől függ** (a paraméterek neve)

A pipeline automatikusan következik a függőségekből – nincs explicit DAG megírás.

````python
import duckdb, random
from datetime import date, timedelta
from dagster import (
    asset, AssetIn, FreshnessPolicy, ConfigurableResource,
    materialize, Definitions
)

# Resource: DuckDB kapcsolat konfigurálása
class DuckDBResource(ConfigurableResource):
    database: str = ':memory:'
    def get_connection(self):
        return duckdb.connect(self.database)

@asset(
    description='Nyers rendelési adatok (szimulált forrás)',
    tags={'layer': 'bronze'},
)
def raw_orders(context, duckdb_resource: DuckDBResource) -> None:
    con   = duckdb_resource.get_connection()
    today = date.today()
    rows  = []
    for i in range(1, 201):
        odate = today - timedelta(days=random.randint(0, 30))
        rows.append({
            'order_id':   i,
            'customer_id': random.randint(1, 50),
            'amount':     round(random.uniform(-100, 50000), 2),  # szándékos negatív
            'status':     random.choice(['completed', 'pending', 'cancelled', None]),
            'order_date': odate.isoformat(),
        })
    con.execute('DROP TABLE IF EXISTS raw_orders')
    con.execute("""
        CREATE TABLE raw_orders AS
        SELECT * FROM (VALUES """ +
        ', '.join(f"({r['order_id']}, {r['customer_id']}, {r['amount']},"
                  f" '{r['status']}', '{r['order_date']}')"
                  for r in rows) +
        """) t(order_id, customer_id, amount, status, order_date)
    """)
    cnt = con.execute('SELECT COUNT(*) FROM raw_orders').fetchone()[0]
    context.log.info(f'raw_orders betöltve: {cnt} sor')

print('raw_orders asset definiálva')
````

````python
@asset(
    ins={'raw_orders': AssetIn()},
    description='Tisztított rendelések – negatív és NULL értékek eltávolítva',
    tags={'layer': 'silver'},
    freshness_policy=FreshnessPolicy(maximum_lag_minutes=60 * 24),
)
def cleaned_orders(context, raw_orders, duckdb_resource: DuckDBResource) -> None:
    con = duckdb_resource.get_connection()
    con.execute('DROP TABLE IF EXISTS cleaned_orders')
    con.execute("""
        CREATE TABLE cleaned_orders AS
        SELECT order_id, customer_id, amount, status,
               CAST(order_date AS DATE) AS order_date
        FROM raw_orders
        WHERE amount > 0
          AND status IS NOT NULL
          AND status != 'cancelled'
    """)
    raw_cnt   = con.execute('SELECT COUNT(*) FROM raw_orders').fetchone()[0]
    clean_cnt = con.execute('SELECT COUNT(*) FROM cleaned_orders').fetchone()[0]
    context.log.info(f'cleaned_orders: {clean_cnt} sor ({raw_cnt - clean_cnt} kiszűrve)')
    context.add_output_metadata({
        'raw_count':   raw_cnt,
        'clean_count': clean_cnt,
        'filtered_pct': round((raw_cnt - clean_cnt) / raw_cnt * 100, 1),
    })

@asset(
    ins={'cleaned_orders': AssetIn()},
    description='Napi bevétel összesítés – gold réteg',
    tags={'layer': 'gold'},
    freshness_policy=FreshnessPolicy(maximum_lag_minutes=60 * 24),
)
def daily_revenue(context, cleaned_orders, duckdb_resource: DuckDBResource) -> None:
    con = duckdb_resource.get_connection()
    con.execute('DROP TABLE IF EXISTS daily_revenue')
    con.execute("""
        CREATE TABLE daily_revenue AS
        SELECT order_date,
               COUNT(*)       AS order_count,
               SUM(amount)    AS total_revenue,
               AVG(amount)    AS avg_order_value
        FROM cleaned_orders
        GROUP BY order_date
        ORDER BY order_date DESC
    """)
    result = con.execute('SELECT COUNT(*), SUM(total_revenue) FROM daily_revenue').fetchone()
    context.log.info(f'daily_revenue: {result[0]} nap, összesen {result[1]:,.0f} összeg')
    context.add_output_metadata({'days_covered': result[0], 'total_revenue': float(result[1])})

print('cleaned_orders és daily_revenue asset-ek definiálva')
````

## 2. Materializáció – asset-ek előállítása

A `materialize()` függvény előállítja az asset-eket a helyes sorrendben,
a függőségi gráf alapján automatikusan: `raw_orders → cleaned_orders → daily_revenue`

````python
# Összes asset materializálása
result = materialize(
    assets=[raw_orders, cleaned_orders, daily_revenue],
    resources={'duckdb_resource': DuckDBResource(database=':memory:')},
)

print(f'Materializáció sikeres: {result.success}')

# Output metadata megtekintése
for event in result.all_node_events:
    if event.event_type_value == 'ASSET_MATERIALIZATION':
        name = event.asset_key.path[-1]
        meta = event.asset_materialization.metadata
        if meta:
            print(f'\n{name} metadata:')
            for k, v in meta.items():
                print(f'  {k}: {v.value}')
````

## 3. Definitions – projekt összefoglaló

A `Definitions` objekt foglalja össze az összes asset-et és resource-t.
Ez az amit a `dagster dev` parancs betölt – ebből épül a Dagit UI.

```bash
# Dagit UI indítása:
dagster dev -f demo3_dagster_assets.py
# → http://localhost:3000
```

````python
defs = Definitions(
    assets=[raw_orders, cleaned_orders, daily_revenue],
    resources={'duckdb_resource': DuckDBResource(database=':memory:')},
)

print('Definitions objektum:')
print(f'  Asset-ek: {[a.key.path[-1] for a in defs.get_all_asset_defs()]}')
print(f'  Resource-ok: {list(defs.resources.keys())}')
````

## 4. Lineage gráf

A Dagit UI Asset Catalog nézetén (http://localhost:3000) vizuálisan látható
az asset-ek közötti függőségi gráf:

```
raw_orders (bronze)
    └── cleaned_orders (silver)  [Freshness: 24h]
            └── daily_revenue (gold)  [Freshness: 24h]
```

Ha a `cleaned_orders` asset 24 óránál régebben materializálódott,
a Dagit `stale` jelzéssel mutatja – ez a Freshness Policy.
